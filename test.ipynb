```python
print(bin(5))
print(bin(1))
c = 5 << 3
print(bin(c))
print(c)
```

```python
print(bin(11))
c = 11 << 1
print(bin(c))
print(c)
m = c >> 1
print(m)
print(bin(m))
```

```python
c = 0b1011
print(f"Исходное:  {c:04b}")

left = ((c << 2) | (c >> 3)) & 0xF
print(f"Влево:     {left:04b}")

right = ((left >> 2) | (left << 3)) & 0xF
print(f"Вправо:    {right:04b}")

print(f"Совпадает? {c == right}")
```
