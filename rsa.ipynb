```python
import math
```

```python
text = "Гркикян Мисак"
b = text.encode("cp1251")
m = int.from_bytes(text.encode("cp1251"), "big") # big - значит, что первый байт в последовательности считается старшим
print(text)
print(text.encode("cp1251"))
print(text.encode("cp1251").hex())
print(list(b))
print(m)
```

```python
# простые числа
p = 1000000007
q = 1000000009
print(p, q)
```

```python
n = p * q
print(n)
```

```python
print(m < n)
```

```python
# функция Эйлера
phi = (p - 1) * (q - 1)
phi
```

```python
e = 65537 # экспонента
```

```python
# с помощью функции проверяем что e и phi имеют один общий делитель - 1 (взаимно простые e и phi)
print(math.gcd(e, phi))
```

```python
d = pow(e, -1, phi)
print(d)
```

```python
print((d * e) % phi == 1)
```

```python
c = pow(m, e, n)
print(c)
```

```python
m_back = pow(c, d, n)
print(m_back)
print(m_back == m)
```

```python
back_to_text = m_back.to_bytes(len(b), "big").decode("cp1251")
print(back_to_text)
```
